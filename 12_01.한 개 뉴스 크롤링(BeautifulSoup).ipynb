```python
# 네이버에서 접속 제한 풀기
# User-Agent 확인: https://www.useragentstring.com/
header = ({"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"
          "AppleWebKit/537.36 (KHTML, like Gecko) Chrome/115.0.0.0 Safari/537.36"})
header
```

## 2.검색어 설정

## 3.URL주소 검색

```python
news_url = f"https://n.news.naver.com/mnews/article/003/0012005629?sid=105"
```

## 4.뉴스 저장


```python
requests = requests.get(news_url, headers = header)
news_html = BeautifulSoup(requests.text, 'html.parser')
```

```python
print(news_html)
```

```python
title = news_html.select_one("#title_area > span").text
title
```
