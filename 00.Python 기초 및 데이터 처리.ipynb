### 6.3 객체지향 클래스

```python
class Cal():

    def __init__(self, x1, x2): # init: 초기값 세팅, self=자기객체를 의미
        self.x1 = x1
        self.x2 = x2

    def add(self):
        result = self.x1 + self.x2
        return result

    def sub(self):
        result = self.x1 - self.x2
        return result
```

```python
# 객체 생성
cal = Cal(3,5)
```

```python
cal.add()
```

```python
cal.sub()
```

## 7.데이터 포멧팅

```python
tot1 = 3.5
print('tot1의 타입은' + type(tot1))
```

```python
tot1 = 3.5
name = '이상철'

# 기본 출력
print('tot1의 타입은', type(tot1), ' 입니다.')
print('tot1: ', tot1)
```

```python
# 문자열 포멧 이용
print('tot1: %0.2f' % tot1)
print('이름: %s' % name)
print('tot1의 값: %0.2f, 이름: %s' % (tot1,name) )
```

```python
# format 이용
print('tot1의 값: {0}'.format(tot1))
print('tot1의 값: {0:0.5f}, 이름: {1:}'.format(tot1, name))
print(f'tot1의 값: {tot1:0.4f}, 이름: {name}')
```
